```python
import pandas as pd
from elasticsearch import Elasticsearch
from datetime import datetime
from pandas import Timestamp
```

```python
from config import ELASTIC_PASSWORD,CERT_PATH

client = Elasticsearch(
    "https://localhost:9200",
    ca_certs=CERT_PATH,
    basic_auth=("elastic",ELASTIC_PASSWORD)  
)

client.info()
```

```python
df = pd.read_csv("df_unido.csv")
df["ciudad"].unique()
```

```python
ciudades_unicas = df["ciudad"].unique()

provincias = {
    "Donosti": "ES-SS",
    "Vitoria": "ES-VI",
    "Bilbao": "ES-BI",
    "Valencia": "ES-V",
    "Madrid": "ES-M",
    "Málaga": "ES-MA",
    "Granada": "ES-GR",
    "Córdoba": "ES-CO",
    "Pamplona": "ES-NA",
}
df['codigo_provincia'] = df['ciudad'].map(provincias)

```

```python
index_name = "provincias"

mapping = {
    "properties": {
        "ciudad": {"type": "text"},
        "codigo_provincia": {"type": "keyword"},
        "timestamp": {"type": "date"}
    }
}

client.indices.delete(index=index_name, ignore_unavailable=True)
client.indices.create(index=index_name, mappings=mapping)
```

```python
for i in range(len(df)):
    data = dict(df.iloc[i])  
    
    doc = {
        "timestamp": str(data["booked_at"]).split(" ")[0], 
        "checkin_time": str(data["checkin_time"]).split(" ")[0],
        "ciudad": data["ciudad"].lower(),
        "codigo_provincia": data["codigo_provincia"],

        "tmed": float(data["tmed"]),  
        "prec": float(data["prec"]), 
        "velmedia": float(data["velmedia"]),  
        "racha": float(data["racha"]),
        "tmin": float(data["tmin"]),  
        "tmax": float(data["tmax"]),  
        "sol": float(data["sol"])
    }
    print(doc)
    client.index(index=index_name, id=i+1, document=doc)
```
