```python
import packages.Preprocesamiento as ppr
import os
import pandas as pd
```

### Carga datos

```python
df_limpio = pd.read_csv('Datos\Transformados\df_limpio.csv')
df_aemet = pd.read_csv('Datos\Transformados\df_AEMET.csv')
```

Cambio de tipo de variables a float

```python
columns_to_replace = ['tmed', 'prec', 'tmin', 'tmax', 'dir', 'velmedia', 'racha', 'sol', 'presMax', 'presMin']

for column in columns_to_replace:
    if df_aemet[column].dtype == 'O':  # 'O' represents object data type (usually strings)
        df_aemet[column] = df_aemet[column].str.replace(',', '.')
```

```python
for column in columns_to_replace:
    df_aemet[column] = df_aemet[column].astype(float)
```

Cambio tipo de variables a fecha

```python
df_limpio['booked_at'] = pd.to_datetime(df_limpio['booked_at'], format = 'mixed')
df_limpio['checkin_time'] = pd.to_datetime(df_limpio['checkin_time'], format = 'mixed')
df_aemet['fecha'] = pd.to_datetime(df_aemet['fecha'], format = 'mixed')
```

### Union de df´s

```python
df_limpio['ciudad'].unique()
```

```python
df_aemet['ciudad'].unique()
```

```python
df_limpio['checkin_time'] = df_limpio['checkin_time'].dt.date
df_aemet['fecha'] = df_aemet['fecha'].dt.date
```

```python
df_merged = df_limpio.merge(df_aemet, left_on = ['ciudad', 'checkin_time'], right_on = ['ciudad', 'fecha'], how = 'left')
df_merged.head()
```

```python
df_merged.to_csv('Datos/Transformados/df_unido.csv', index=False)
```
